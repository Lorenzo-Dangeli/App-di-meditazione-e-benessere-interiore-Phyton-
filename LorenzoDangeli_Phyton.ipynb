```python
#Sempre all'interno dell'app ho creato una lista di obiettivi che l'utente può scegliere durante la sessione di meditazione
OBIETTIVI_DISPONIBILI = [
    "Ridurre lo stress",
    "Migliorare lo stato d'animo",
    "Ascoltare se stessi"
]
```

```python
#Ho creato una funzione per la selezione interattiva dell'obiettivo
def seleziona_obiettivo_sicuro():
    print("\nSeleziona l'obiettivo per la sessione di oggi:")
    for i, obj in enumerate(OBIETTIVI_DISPONIBILI, 1):
        print(f"{i}. {obj}")
        
    while True:
        try:
            scelta = int(input("Inserisci il numero dell'obiettivo scelto: "))
            if 1 <= scelta <= len(OBIETTIVI_DISPONIBILI):
                obiettivo_scelto = OBIETTIVI_DISPONIBILI[scelta - 1]
                print(f"Ottima scelta! Oggi lavoreremo su: {obiettivo_scelto}")
                return obiettivo_scelto
            else:
                print(f"Errore: inserisci un numero valido compreso tra 1 e {len(OBIETTIVI_DISPONIBILI)}.")
        except ValueError:
            print("Errore: devi inserire un valore numerico valido!")
```

```python
# 1. Creo una funzione per registrare una nuova sessione
def registra_sessione(collezione, utente, musica, durata, ambiente, obiettivo):
    sessione_attuale = {
        "utente": utente.nome_utente,
        "musica": musica,
        "durata": durata,
        "ambiente": ambiente,
        "obiettivo": obiettivo
    }
    collezione.append(sessione_attuale)
    print("Sessione registrata con successo!")

# 2. Funzione per stampare l'elenco completo
def mostra_sessioni(collezione):
    print("---ELENCO SESSIONI REGISTRATE---")
    if not collezione:
        print("Nessuna sessione presente nella collezione.")
        return
    for i, s in enumerate(collezione, 1):
        print(f"Sessione {i} Utente: {s['utente']} | Obiettivo: {s['obiettivo']} | Durata: {s['durata']} min | Ambiente: {s['ambiente']}")
        print("-" * 30)
```

```python
#Funzione per calcolare metriche e distribuzione per categoria
def calcola_statistiche(collezione):
  if not collezione:
    print("Nessuna sessione presente nella collezione.")
    return

  totale_sessioni = len(collezione)
  durata_totale = sum(s["durata"] for s in collezione)
  durata_media = durata_totale / totale_sessioni

  conteggio_obiettivi = {}
  for s in collezione:
    obj = s["obiettivo"]
    conteggio_obiettivi[obj] = conteggio_obiettivi.get(obj, 0) + 1

  print("--- STATISTICHE AVANZATE DI MEDITAZIONE ---")
  print(f"Numero totale di sessioni: {totale_sessioni}")
  print(f"Minuti totali dedicati al benessere: {durata_totale} minuti")
  print(f"Durata media per sessione: {durata_media:.1f} minuti")
  print("\nDistribuzione per obiettivo:")
  for obiettivo, conteggio in conteggio_obiettivi.items():
    print(f"• {obiettivo}: {conteggio} sessioni")
  print("-" * 40)

#Filtraggio sessioni con criteri opzionali
def filtra_sessioni(collezione, obiettivo=None, ambiente=None, durata_max=None):
  print("--- FILTRAGGIO AVANZATO SESSIONI ---")
  risultati = []

  for s in collezione:
    match = True
    if obiettivo and s["obiettivo"].lower() != obiettivo.lower():
      match = False
    if ambiente and s["ambiente"].lower() != ambiente.lower():
      match = False
    if durata_max is not None and s["durata"] > durata_max:
      match = False

    if match:
      risultati.append(s)

  if risultati:
    print(f"Trovate {len(risultati)} sessioni corrispondenti ai criteri:")
    for s in risultati:
      print(
          f" -> Utente: {s['utente']} | Obiettivo: {s['obiettivo']} | Durata:"
          f" {s['durata']} min | Ambiente: {s['ambiente']}"
      )
  else:
    print("Nessuna sessione trovata con i criteri specificati.")
  print("-" * 40)
```

```python
# Carico le 4 sessioni di esempio nella collezione
carica_sessioni_esempio()

# Qui calcolo e stampo le statistiche aggiornate
calcola_statistiche(collezione_sessione)
```
